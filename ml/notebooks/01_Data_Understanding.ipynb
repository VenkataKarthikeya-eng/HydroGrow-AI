# 01. Data Understanding & Dataset Audit

## Evaluation Dataset Metadata
- **Scope**: Exploratory audit of all four HydroGrow AI model datasets.
- **Data Partition**: Complete Dataset Audits (Raw + Processed).

| Dataset | Dataset Source Location | Evaluated Samples | Partition Type | Class / Target Schema |
| :--- | :--- | :--- | :--- | :--- |
| **Tabular Yield** | `data/processed/final_ml_dataset.csv` | 218 records | Full Dataset Audit | Target: `target_total_weight_g` (34 features) |
| **Crop Validator** | `data/crop_validator_dataset/` | 1,500 images | Full Dataset Audit | `lettuce_leaf`, `other_plant_leaf`, `non_leaf` |
| **Growth Stage & Day** | `data/growth_dataset/` | 300 images | Full Dataset Audit | `Month1`, `Month2`, `Month3` (Days 1–30) |
| **Nutrient Deficiency** | `data/nutrient_dataset/` | 1,200 images | Full Dataset Audit | `healthy`, `nitrogen`, `phosphorus`, `potassium` |

---


```python
import os
import pandas as pd
import glob
from PIL import Image

PROJECT_ROOT = "../.."

# 1. Tabular Yield Dataset Audit
yield_csv = os.path.join(PROJECT_ROOT, "data", "processed", "final_ml_dataset.csv")
if os.path.exists(yield_csv):
    df_yield = pd.read_csv(yield_csv)
    print("=== 1. Fresh Weight Yield Dataset Summary ===")
    print(f"Dataset Source: {yield_csv}")
    print(f"Total Records: {len(df_yield)}")
    print(f"Target Feature: 'target_total_weight_g'")
    print(f"Feature Count: {len(df_yield.columns)}")
    print("\nFirst 5 Records:")
    print(df_yield[['experiment', 'system', 'target_total_weight_g', 'water_ph_mean', 'water_ec_mean']].head().to_string())

# 2. Crop Validation Vision Dataset Audit
crop_val_dir = os.path.join(PROJECT_ROOT, "data", "crop_validator_dataset")
print("\n=== 2. Crop Validation Vision Dataset Summary ===")
print(f"Dataset Source: {crop_val_dir}")
for cls in ["lettuce_leaf", "other_plant_leaf", "non_leaf"]:
    cls_path = os.path.join(crop_val_dir, cls)
    count = len(glob.glob(os.path.join(cls_path, "*.*"))) if os.path.exists(cls_path) else 0
    print(f"Class '{cls}': {count} images")

# 3. Growth Stage & Day Vision Dataset Audit
growth_dir = os.path.join(PROJECT_ROOT, "data", "growth_dataset")
print("\n=== 3. Growth Stage & Day Vision Dataset Summary ===")
print(f"Dataset Source: {growth_dir}")
for month in ["Month1", "Month2", "Month3"]:
    m_path = os.path.join(growth_dir, month)
    count = len(glob.glob(os.path.join(m_path, "**", "*.*"), recursive=True)) if os.path.exists(m_path) else 0
    print(f"Folder '{month}': {count} images")

# 4. Nutrient Deficiency Vision Dataset Audit
nutrient_dir = os.path.join(PROJECT_ROOT, "data", "nutrient_dataset")
print("\n=== 4. Nutrient Deficiency Vision Dataset Summary ===")
print(f"Dataset Source: {nutrient_dir}")
for cls in ["healthy", "nitrogen_deficiency", "phosphorus_deficiency", "potassium_deficiency"]:
    cls_path = os.path.join(nutrient_dir, cls)
    count = len(glob.glob(os.path.join(cls_path, "*.*"))) if os.path.exists(cls_path) else 0
    print(f"Class '{cls}': {count} images")
```
