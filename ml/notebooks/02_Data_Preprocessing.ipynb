## Preprocessing Specifications
- **Vision Pipeline**: Load RGB images $\rightarrow$ bilinear resize to $(224, 224, 3)$ $\rightarrow$ float32 normalization to $[0.0, 1.0]$ $\rightarrow$ batch dimension expansion $(1, 224, 224, 3)$.
- **Tabular Pipeline**: Select 34 sensor features $\rightarrow$ missing value check $\rightarrow$ StandardScaler scaling $\rightarrow$ $80\%/20\%$ train/test partition.
